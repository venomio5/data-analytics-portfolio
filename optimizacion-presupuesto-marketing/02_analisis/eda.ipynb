```python
import pandas as pd
```

```python
campanas_df = pd.read_csv(r"..\01_datos\processed\campanas.csv")
conversiones_df = pd.read_csv(r"..\01_datos\processed\conversiones.csv")
transacciones_df = pd.read_csv(r"..\01_datos\processed\transacciones.csv")
```

```python
campanas_df
```

```python
campanas_df = campanas_df.rename(columns={'gasto_mxn': 'gastos', 'valor_conv_mxn': 'valor_conv'})
metricas = ['impresiones', 'clics', 'conversiones', 'gastos', 'valor_conv']
```

```python
canal_df = campanas_df.groupby(['canal'])[metricas].sum().reset_index()

canal_df['CTR'] = ((canal_df['clics'] / canal_df['impresiones'])*100).fillna(0).round(1)
canal_df['CVR'] = ((canal_df['conversiones'] / canal_df['clics'])*100).fillna(0).round(1)
canal_df['CAC'] = (canal_df['gastos'] / canal_df['conversiones']).fillna(0).round(1)
canal_df['ROAS'] = (canal_df['valor_conv'] / canal_df['gastos']).fillna(0).round(1)
canal_df['CPM'] = (canal_df['gastos'] / canal_df['impresiones'] * 1000).fillna(0).round(1)

canal_df = canal_df[['canal', 'ROAS', 'CAC', 'CPM', 'CTR', 'CVR']]

canal_df
```

```python
tipo_campana_df = campanas_df.groupby(['tipo_campana', 'canal'])[metricas].sum().reset_index()

tipo_campana_df['CTR'] = ((tipo_campana_df['clics'] / tipo_campana_df['impresiones'])*100).fillna(0).round(1)
tipo_campana_df['CVR'] = ((tipo_campana_df['conversiones'] / tipo_campana_df['clics'])*100).fillna(0).round(1)
tipo_campana_df['CAC'] = (tipo_campana_df['gastos'] / tipo_campana_df['conversiones']).fillna(0).round(1)
tipo_campana_df['ROAS'] = (tipo_campana_df['valor_conv'] / tipo_campana_df['gastos']).fillna(0).round(1)
tipo_campana_df['CPM'] = (tipo_campana_df['gastos'] / tipo_campana_df['impresiones'] * 1000).fillna(0).round(1)
tipo_campana_df['Inversion'] = (tipo_campana_df['gastos']).fillna(0).round(1)

tipo_campana_df = tipo_campana_df[['tipo_campana', 'canal', 'Inversion', 'ROAS', 'CAC', 'CPM', 'CTR', 'CVR']]

tipo_campana_df
```

```python
campana_id_df = campanas_df.groupby(['campana_id', 'tipo_campana'])[metricas].sum().reset_index()

campana_id_df['CTR'] = ((campana_id_df['clics'] / campana_id_df['impresiones'])*100).fillna(0).round(1)
campana_id_df['CVR'] = ((campana_id_df['conversiones'] / campana_id_df['clics'])*100).fillna(0).round(1)
campana_id_df['CAC'] = (campana_id_df['gastos'] / campana_id_df['conversiones']).fillna(0).round(1)
campana_id_df['ROAS'] = (campana_id_df['valor_conv'] / campana_id_df['gastos']).fillna(0).round(1)
campana_id_df['CPM'] = (campana_id_df['gastos'] / campana_id_df['impresiones'] * 1000).fillna(0).round(1)
campana_id_df['CPC'] = (campana_id_df['gastos'] / campana_id_df['clics']).fillna(0).round(2)
campana_id_df['Inversion'] = (campana_id_df['gastos']).fillna(0).round(1)

campana_id_df = campana_id_df[['campana_id', 'tipo_campana', 'Inversion', 'ROAS', 'CAC', 'CPM', 'CTR', 'CPC', 'CVR']]

campana_id_df
```
